# 🏥 Auditoria Macrocontábil e Estatística: Impacto de Estrangeiros no SUS (2023–2025)

**Objetivo:** Demonstrar quantitativamente, sob a ótica da Ciência de Dados e da Contabilidade Aplicada ao Setor Público, que o volume de atendimentos a estrangeiros no SUS configura uma **variação residual** (< 0,03%), absorvida integralmente pela **margem de segurança** orçamentária dos orçamentos consolidados (União, Estados e Municípios via SIOPS).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Configuração visual dos gráficos
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['font.size'] = 11
```

## 1. Consolidação da Série Histórica Nacional e Orçamento Tripartite (2023–2025)
Cruzamento entre os atendimentos a cidadãos estrangeiros (residentes e não residentes), a produção total do SUS (~2,8 bilhões de procedimentos) e o orçamento consolidado (Federal, Estadual e Municipal).

```python
dados_nacionais = {
    'Ano': [2023, 2024, 2025],
    'Orcamento_Federal_MS': [155000000000.00, 162000000000.00, 170000000000.00],
    'Orcamento_Estadual_SES': [95000000000.00, 100000000000.00, 105000000000.00],
    'Orcamento_Municipal_SMS': [100000000000.00, 103000000000.00, 105000000000.00],
    'Atendimentos_Estrangeiros': [1143223, 870000, 775619],
    'Producao_Total_SUS': [2800000000, 2800000000, 2800000000],
    'Status_Dado': ['Oficial (Pico)', 'Estimado (Acomodação)', 'Oficial (Estabilização)']
}

df_sus = pd.DataFrame(dados_nacionais)
df_sus['Orcamento_Total_SUS'] = df_sus['Orcamento_Federal_MS'] + df_sus['Orcamento_Estadual_SES'] + df_sus['Orcamento_Municipal_SMS']
df_sus['Margem_Seguranca_2pct'] = df_sus['Orcamento_Total_SUS'] * 0.02
df_sus['Representatividade_Percentual (%)'] = (df_sus['Atendimentos_Estrangeiros'] / df_sus['Producao_Total_SUS']) * 100
df_sus['Gasto_Estrangeiros_Reais'] = df_sus['Orcamento_Total_SUS'] * (df_sus['Atendimentos_Estrangeiros'] / df_sus['Producao_Total_SUS'])
df_sus['Consumo_Margem_Seguranca (%)'] = (df_sus['Gasto_Estrangeiros_Reais'] / df_sus['Margem_Seguranca_2pct']) * 100

display(df_sus[['Ano', 'Orcamento_Total_SUS', 'Margem_Seguranca_2pct', 'Atendimentos_Estrangeiros', 'Representatividade_Percentual (%)', 'Gasto_Estrangeiros_Reais', 'Consumo_Margem_Seguranca (%)']].style.format({
    'Orcamento_Total_SUS': 'R$ {:,.2f}',
    'Margem_Seguranca_2pct': 'R$ {:,.2f}',
    'Atendimentos_Estrangeiros': '{:,.0f}',
    'Representatividade_Percentual (%)': '{:.4f}%',
    'Gasto_Estrangeiros_Reais': 'R$ {:,.2f}',
    'Consumo_Margem_Seguranca (%)': '{:.2f}%'
}))
```

## 2. Visualização Gráfica da Evolução de Atendimentos

```python
fig, ax = plt.subplots(figsize=(9, 5))

anos = df_sus['Ano'].astype(str)
valores = df_sus['Atendimentos_Estrangeiros'] / 1000

barras = ax.bar(anos, valores, color=['#1b4965', '#62b6cb', '#5fa8d3'], width=0.5, edgecolor='black', alpha=0.9)
ax.plot(anos, valores, color='#e63946', marker='o', linewidth=2.5, label='Tendência de Acomodação (-32,15%)')

for barra in barras:
    altura = barra.get_height()
    ax.annotate(f'{altura:,.1f}k',
                xy=(barra.get_x() + barra.get_width() / 2, altura),
                xytext=(0, 8),
                textcoords="offset points",
                ha='center', va='bottom', fontweight='bold')

ax.set_title('Evolução de Atendimentos a Estrangeiros no SUS (2023–2025)', fontsize=14, fontweight='bold', pad=15)
ax.set_ylabel('Volume de Atendimentos (em milhares)', fontweight='bold')
ax.set_ylim(0, 1350)
ax.legend()
plt.tight_layout()
plt.show()
```

## 3. Conclusão Técnica e Contábil
1. **Escala Nacional:** O atendimento a estrangeiros representa apenas **0,0277%** da produção do SUS em 2025, com retração acumulada expressiva.
2. **Auditoria Orçamentária:** O impacto financeiro estimado de **R$ 105,26 milhões** consome apenas **1,39%** de toda a Margem de Segurança Orçamentária nacional (R$ 7,6 bilhões), confirmando a hipótese de variação residual.
